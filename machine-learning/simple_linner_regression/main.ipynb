# Practice: Understanding Simple Linear Regression

## Importing Libraries We Need

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Generating Data

The relationship between x and y is:

**y = 2x - 1**

```python
x = np.array([1,2,3,4]).reshape(-1,1)
y = np.array([1,3,5,7]).reshape(-1,1)
```

```python
# plot

plt.plot(x,y,"--r")
plt.show()
```

```python
# model

from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x,y)
```

```python
# y = m * x + c

m = model.coef_[0]
c = model.intercept_

print(m,c)
```

```python
plt.plot(x,y,"*")
plt.plot(x,model.predict(x),"red")
plt.show()
```

```python
result = model.predict([[5]])
result
```

# Part 2: Regression with Real-World Data

In the previous example, the relationship between x and y was completely linear, so the model could predict the output almost perfectly.

In real-world problems, data usually contains noise and the relationship between input and output is not always simple.

In this example, we will train a regression model to predict salary based on years of experience and see that predictions are not always exactly equal to the real values.

```python
years_of_experience = np.array([0,1,2,3,4,6,8,10]).reshape(-1,1)
salaries = np.array([30000,40000,55000,60000,70000,80000,85000,87000])
```

```python
plt.plot(years_of_experience,salaries,'*')
plt.show()
```

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(max_depth=4)

model.fit(years_of_experience,salaries)
y_predict = model.predict(years_of_experience)

plt.plot(years_of_experience, salaries, '*')
plt.plot(years_of_experience, y_predict)
plt.show()
```

```python
x_test = np.linspace(0, 10, 100).reshape(-1,1)

y_test = model.predict(x_test)

plt.scatter(years_of_experience, salaries)
plt.plot(x_test, y_test)
plt.show()
```
