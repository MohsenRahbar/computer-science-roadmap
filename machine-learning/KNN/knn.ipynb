## K-Nearest Neighbors (KNN)

K-Nearest Neighbors is a supervised machine learning
algorithm used for classification and regression.

In this example, we use KNN for classification.
The algorithm predicts the class of a sample based
on the majority class among its nearest neighbors.

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('./diabetes.csv')
df.head()
```

```python
df.shape
```

```python
df['Outcome'].value_counts()
```

```python
x = df.drop('Outcome',axis=1)
y = df['Outcome']
```

```python
x.head()
```

```python
y.head()
```

```python
x = np.array(x)
y = np.array(y)
```

```python
x
```

```python
y
```

# Normalize

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x = scaler.fit_transform(x)
```

```python
x
```

# test/train

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
# 20 percent(0.2) for test and 80% for train
```

```python
x_train.shape,y_train.shape
# 80 %
```

```python
x_test.shape,y_test.shape
# 20 %
```

## Train-Test Split

We split the dataset into two subsets:

- Training set: 80% of the data
- Testing set: 20% of the data

The training set is used to fit the model, while
the testing set is used to evaluate its performance.

# KNN

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=4)
knn.fit(x_train,y_train)
```

```python
y_pred_train = knn.predict(x_train)
y_pred_test = knn.predict(x_test)
```

### Model Evaluation

We evaluate the trained model using the following metrics:

- Accuracy
- Precision
- Recall
- Confusion Matrix

```python
from sklearn .metrics import accuracy_score

acc_train = accuracy_score(y_train,y_pred_train)
acc_test = accuracy_score(y_test,y_pred_test)

acc_train,acc_test
```

```python
acc_train*100,acc_test*100
```

```python
from sklearn.metrics import confusion_matrix, precision_score ,recall_score
xx=confusion_matrix(y_test,y_pred_test)
p = precision_score(y_test,y_pred_test)
r = recall_score(y_test,y_pred_test)

xx
```

```python
p*100
```

```python
r*100
```

### Observation

The evaluation metrics help us understand how well
the model performs on both training and testing data.

## Experimenting with Different K Values

In this experiment, we increase the number of neighbors
(K) from 4 to 8 to observe how changing the K value
affects the model's classification performance.

We then compare the evaluation results of both models
to understand the impact of the number of neighbors.

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=8)
knn.fit(x_train,y_train)
```

```python
y_pred_train = knn.predict(x_train)
y_pred_test = knn.predict(x_test)
```

```python
from sklearn .metrics import accuracy_score

acc_train = accuracy_score(y_train,y_pred_train)
acc_test = accuracy_score(y_test,y_pred_test)

acc_train,acc_test
```

```python
from sklearn.metrics import confusion_matrix, precision_score ,recall_score
confusion_matrix(y_test,y_pred_test)
```

```python
p = precision_score(y_test,y_pred_test)
r = recall_score(y_test,y_pred_test)
r*100
```

```python
p*100
```
