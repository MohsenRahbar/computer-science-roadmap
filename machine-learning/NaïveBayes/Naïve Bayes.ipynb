# Normalize

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x = scaler.fit_transform(x)
```

```python
x
```

# test/train

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
# 20 percent(0.2) for test and 80% for train
```

```python
x_train.shape,y_train.shape
# 80 %
```

```python
x_test.shape,y_test.shape
# 20 %
```

# Naïve Bayes

Naïve Bayes is a supervised machine learning algorithm
based on Bayes' Theorem.

It uses probability to predict the class of a given sample.
The algorithm assumes that features are conditionally
independent given the class.

```python
from sklearn.naive_bayes import GaussianNB

modle = GaussianNB()
modle.fit(x_train,y_train)
```

## Gaussian Naïve Bayes

GaussianNB is a Naïve Bayes classifier that assumes
the features follow a Gaussian (normal) distribution
within each class.

We create the model and train it using the training data.

```python
y_pred_train = modle.predict(x_train)
y_pred_test = modle.predict(x_test)
```

## Model Prediction

After training the model, we use it to predict the
classes of both training and testing datasets.

This allows us to evaluate the model's performance
on both datasets.

```python
from sklearn.metrics import accuracy_score

acc_train = accuracy_score(y_true=y_train,y_pred=y_pred_train)
acc_test = accuracy_score(y_true=y_test,y_pred=y_pred_test)

acc_train,acc_test
```

```python
from sklearn.metrics import confusion_matrix, precision_score ,recall_score

confusion_matrix(y_test,y_pred_test)
```

```python
p = precision_score(y_test,y_pred_test)
p*100
```

```python
r = recall_score(y_test, y_pred_test)
r*100
```
